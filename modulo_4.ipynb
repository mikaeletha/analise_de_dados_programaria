### Módulo 4 - ESTATISTICA

Faça a média da idade de quem respondeu ser do gênero masculino, assim como as estatísticas da coluna de salário

```python
dados = pd.read_excel('data/planilha_modulo3.xlsx')
```

```python
masculino = dados[dados['GENERO'] == 'Masculino']
masculino['IDADE'].mean()
```

```python
salario = masculino['SALARIO']
```

```python
# média
salario.mean()
```

```python
# mediana
salario.median()
```

```python
# moda
salario.mode()
```

```python
# desvio padrão
salario.std()
```

```python
# menor valor
salario.min()
```

```python
# maior valor
salario.max()
```
