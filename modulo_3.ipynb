### Módulo 3 - FILTRO

Que tal fazer um filtro para identificar pessoas que se declararam amarelas e outro de pessoas abaixo de 40 anos? E melhor, um filtro que misture ambos?

```python
dados.columns
```

```python
dados.groupby('COR/RACA/ETNIA', dropna=False)['ID'].nunique()
```

```python
amarela = dados[dados['COR/RACA/ETNIA']=='Amarela']
amarela
```

```python
idade_40 = dados[dados['IDADE']<40]
idade_40
```

```python
amarela_40 = dados[(dados['COR/RACA/ETNIA']=='Amarela') & (dados['IDADE']<40)]
amarela_40
```

### Módulo 4 - ESTATISTICA

Faça a média da idade de quem respondeu ser do gênero masculino, assim como as estatísticas da coluna de salário

```python
masculino = dados[dados['GENERO'] == 'Masculino']
masculino['IDADE'].mean()
```

```python
salario = masculino['SALARIO']
```

```python
# média
salario.mean()
```

```python
# mediana
salario.median()
```

```python
# moda
salario.mode()
```

```python
# desvio padrão
salario.std()
```

```python
# menor valor
salario.min()
```

```python
# maior valor
salario.max()
```
