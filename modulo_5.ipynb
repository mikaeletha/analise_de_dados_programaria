Correlação entre nível de ensino e gênero

```python
from scipy.stats import chi2_contingency
```

```python
def cramer(col_1, col_2):
    tabela_cruzada = np.array(
        pd.crosstab(col_1, col_2)
    )
    chi_2 = chi2_contingency(tabela_cruzada)[0]
    soma = np.sum(tabela_cruzada)
    min = np.min(tabela_cruzada.shape)-1
    cramer = np.sqrt(chi_2/(soma*min))

    return cramer
```

```python
cramer(dados['NIVEL DE ENSINO'], dados['GENERO'])
```
